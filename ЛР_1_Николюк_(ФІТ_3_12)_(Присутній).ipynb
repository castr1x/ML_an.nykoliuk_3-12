Дано матриці:

```text
B = [[2, 0],

```

Обчисліть:
1. поелементний добуток `A * B`;
2. матричний добуток `A @ B`.

Поясніть у коментарі до коду, чим відрізняються ці операції.

````python
# Код студента
A = np.array([[1, 2],
     [3, 4]])

B = np.array([[2, 0],
     [1, 2]])

# поелементний добуток A * B - множить числа на однакових місцях, матриці мають бути однакового розміру
# матричний добуток A @ B - множить рядок на стовпець, кількість рядків 1 матриці має дорівнювати кількрості стовпців 2 матриці

result = [A*B,"\n",A@B]

for i in result:
  print(i)
````

### Завдання 28
Для матриці:

```text
A = [[4, 7],
     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
# Код студента
A = np.array(
    [
        [4, 7],
        [2, 6]
    ]
  )

det = np.linalg.det(A)
inv = np.linalg.inv(A)

check = np.round(A @ inv)

result = [A, np.round(det), inv, check]

for i in result:
  print(i,"\n")
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array(
    [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
    )


mu = np.mean(A, axis = 1)
sigma = np.std(A, axis = 1)

z = (A-mu)/sigma
z
````

```text
         [200, 180, 210, 230]]
```

2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
mtx = np.random.randint(1,21,size = 25).reshape(5,5)

d_1 = np.linalg.diagonal(mtx, offset=0)
d_2 = np.linalg.diagonal(np.fliplr(mtx), offset = 0)

copy = mtx.copy()

copy[copy < copy.mean()] = 0

result = [mtx, d_1.sum(), d_2.sum(), np.max(mtx, axis = 1), np.min(mtx, axis = 0), copy]

for i in result:
  print(i,'\n')
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
mtx = np.arange(1,17,1).reshape(4,4)

norms = np.linalg.norm(mtx, axis = 1, keepdims=True) #True поветиає масив (4,1), а не (4) як за замовчуванням
mtx_norm  = mtx/norms

result = [mtx, norms, mtx_norm]

for i in result:
  print(i,'\n')
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array(
    [
        [1,2,1],
        [2,1,-1],
        [3,-1,2]
    ]

)

b = np.array([9,8,3])

x = np.linalg.solve(A,b)

result = [A,b,x,A@x]

for i in result:
  print(i, '\n')
````
